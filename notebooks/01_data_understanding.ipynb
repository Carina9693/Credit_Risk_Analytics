```python
analysis_results = {
    'Credit Score': credit_analysis,
    'Income': income_analysis,
    'DTI Ratio': dti_analysis,
    'Loan Amount': loan_analysis,
    'Employment Type': employment_analysis,
    'Employment Duration': duration_analysis,
    'Interest Rate': interest_analysis,
    'Loan Purpose': purpose_analysis,
    'Marital Status': marital_analysis,
    'Mortgage Status': mortgage_analysis,
    'Co-signer Status': cosigner_analysis
}

summary = []

for factor, result in analysis_results.items():

    lowest = result['Default_Rate'].idxmin()
    highest = result['Default_Rate'].idxmax()

    summary.append({
        'Factor': factor,
        'Lowest Risk Group': str(lowest),
        'Lowest Default Rate (%)': round(
            result.loc[lowest, 'Default_Rate'], 2
        ),
        'Highest Risk Group': str(highest),
        'Highest Default Rate (%)': round(
            result.loc[highest, 'Default_Rate'], 2
        )
    })

eda_summary = pd.DataFrame(summary)

eda_summary
```

```python
query9 = """
WITH Risk_Summary AS (
    SELECT
        CASE
            WHEN CreditScore < 500
                 AND Income < 30000
                 AND DTIRatio > 0.5
            THEN 'High Risk'
            WHEN CreditScore < 600
                 AND Income < 60000
            THEN 'Medium Risk'
            ELSE 'Lower Risk'
        END AS Risk_Category,
        COUNT(*) AS Total_Borrowers,
        SUM("Default") AS Total_Defaulters
    FROM loans
    GROUP BY Risk_Category
),
Portfolio_Summary AS (
    SELECT
        100.0 * SUM("Default") / COUNT(*) AS Overall_Default_Rate
    FROM loans
)
SELECT
    r.Risk_Category,
    r.Total_Borrowers,
    r.Total_Defaulters,
    ROUND(100.0 * r.Total_Defaulters / r.Total_Borrowers, 2)
        AS Default_Rate,
    ROUND(p.Overall_Default_Rate, 2) AS Overall_Default_Rate
FROM Risk_Summary r
CROSS JOIN Portfolio_Summary p
WHERE 100.0 * r.Total_Defaulters / r.Total_Borrowers
      > p.Overall_Default_Rate
ORDER BY Default_Rate DESC;
"""

pd.read_sql_query(query9, conn)
```

```python
# 1. Overall loan summary
overall_summary = pd.read_sql_query("""
SELECT
    COUNT(*) AS Total_Loans,
    SUM("Default") AS Total_Defaulters,
    COUNT(*) - SUM("Default") AS Non_Defaulters,
    ROUND(100.0 * SUM("Default") / COUNT(*), 2) AS Default_Rate
FROM loans;
""", conn)

overall_summary.to_csv("../outputs/overall_loan_summary.csv", index=False)


# 2. Loan purpose analysis
purpose_summary = pd.read_sql_query("""
SELECT
    LoanPurpose,
    COUNT(*) AS Total_Loans,
    SUM("Default") AS Total_Defaulters,
    ROUND(AVG(LoanAmount), 2) AS Average_Loan_Amount,
    ROUND(100.0 * SUM("Default") / COUNT(*), 2) AS Default_Rate
FROM loans
GROUP BY LoanPurpose;
""", conn)

purpose_summary.to_csv("../outputs/loan_purpose_summary.csv", index=False)


# 3. Credit score analysis
credit_summary = pd.read_sql_query("""
SELECT
    CASE
        WHEN CreditScore < 500 THEN 'Below 500'
        WHEN CreditScore < 600 THEN '500-599'
        WHEN CreditScore < 700 THEN '600-699'
        WHEN CreditScore < 800 THEN '700-799'
        ELSE '800+'
    END AS Credit_Score_Group,
    COUNT(*) AS Total_Borrowers,
    SUM("Default") AS Total_Defaulters,
    ROUND(100.0 * SUM("Default") / COUNT(*), 2) AS Default_Rate
FROM loans
GROUP BY Credit_Score_Group;
""", conn)

credit_summary.to_csv("../outputs/credit_score_summary.csv", index=False)


print("All three summaries exported successfully!")
```
